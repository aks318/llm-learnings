```python
from langgraph.graph import StateGraph, START, END
from typing import TypedDict
from langchain_ollama import ChatOllama
```

```python
llm = ChatOllama(model="llama3.2:1b")
```

```python
class BlogState(TypedDict):
    title: str
    outline: str
    blog: str
```

```python
def getOutline(state: BlogState) -> BlogState:
    title = state['title']

    prompt = f"Provide the outline of the given title: {title}"

    res1 = llm.invoke(prompt)
    state['outline'] = res1.content

    return state
```

```python
def getBlog(state: BlogState) -> BlogState:
    outline = state['outline']

    prompt = f"Provide the blog post of the given outline: \n {outline}"

    res2 = llm.invoke(prompt)
    state['blog'] = res2.content

    return state
```

```python
graph = StateGraph(BlogState)
graph.add_node("getOutline", getOutline)
graph.add_node("getBlog", getBlog)

graph.add_edge(START, "getOutline")
graph.add_edge("getOutline", "getBlog")
graph.add_edge("getBlog", END)
```

```python
workflow = graph.compile()
final_res = workflow.invoke({"title" : "Rise of AI in india"})
```
